2. Imagine you have two 3x3 matrices represented as lists of lists. How would you write the standard Python code to add them together, number by number?

```python
A=[
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
B=[
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
result=[]
for i in range(3):
    row=[]
    for j in range(3):
        row.append(A[i][j]+B[i][j])
    result.append(row)
print(result)
```
